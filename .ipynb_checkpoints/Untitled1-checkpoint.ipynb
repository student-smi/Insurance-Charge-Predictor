```python
df_cleaned_data = df.copy()
```

```python
df_cleaned_data.head()
```

```python
df_cleaned_data.shape
```

```python
df_cleaned_data.drop_duplicates(inplace=True)
```

```python
df_cleaned_data.shape
```

```python
df_cleaned_data['smoker'].value_counts()
```

```python
df_cleaned_data['sex'] = df_cleaned_data['sex'].map({'male' : 0 , "female" : 1})
```

```python
df_cleaned_data.head()
```

```python
df_cleaned_data['smoker'] = df_cleaned_data['smoker'].map({'no' : 0 , "yes" : 1})
df_cleaned_data.head()
```

```python
df_cleaned_data.rename(columns={
    'sex': "is_female",
    'smoker' : 'is_smoker'
} , inplace=True)
```

```python
df_cleaned_data.head()
```

```python
df_cleaned_data = pd.get_dummies(df_cleaned_data, columns=['region'] , drop_first=True)
```

```python
df_cleaned_data.head()
```

```python
df_cleaned_data = df_cleaned_data.astype(int)
```

```python
df_cleaned_data
```

```python
df_cleaned_data['bmi_category'] = pd.cut(
    df_cleaned_data['bmi'],
    bins=[0, 18.5, 24.9, 29.9, float('inf')],
    labels=['Underweight', 'Normal', 'Overweight', 'Obese']
)
df_cleaned_data.head()
```

```python
df_cleaned_data = pd.get_dummies( df_cleaned_data, columns=['bmi_category'] , drop_first=True)
```

```python
df_cleaned_data = df_cleaned_data.astype(int)
```

```python
df_cleaned_data.head()
```

```python
from sklearn.preprocessing import StandardScaler
colm = ['age', 'bmi' , 'children']
slace = StandardScaler()
df_cleaned_data[colm]= slace.fit_transform(df_cleaned_data[colm])

```

```python
df_cleaned_data.head()
```

```python
from scipy.stats import pearsonr

# ----------------------------------
# Pearson Correlation Calculation
# ----------------------------------

# List of features to check against target
selected_features = [
    'age', 'bmi', 'children', 'is_female', 'is_smoker',
    'region_northwest', 'region_southeast', 'region_southwest',
    'bmi_category_Normal', 'bmi_category_Overweight', 'bmi_category_Obese'
]

correlations = {
    feature: pearsonr(df_cleaned_data[feature], df_cleaned_data['charges'])[0]
    for feature in selected_features
}
correlation_df = pd.DataFrame(list(correlations.items()), columns=['Feature', 'Pearson Correlation'])
correlation_df.sort_values(by='Pearson Correlation', ascending=False)
```

```python
final_df = df_cleaned_data[['age', 'is_female', 'bmi', 'children', 'is_smoker', 'charges','region_southeast','bmi_category_Obese']]
```
